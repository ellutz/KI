#### Minimax
Der Minimax-Algorithmus ist ein Entscheidungsalgorithmus, der häufig in Spielen wie Schach oder Tic-Tac-Toe verwendet wird, um optimale Züge zu berechnen. Er wird eingesetzt, wenn zwei Spieler gegeneinander spielen und beide versuchen, das beste Ergebnis für sich zu erzielen. Die Grundidee des Algorithmus ist, dass jeder Spieler versucht, seinen eigenen Gewinn zu maximieren, während er gleichzeitig den Gegner daran hindert, zu gewinnen.

<img src="https://miro.medium.com/v2/resize:fit:720/format:webp/1*GaJ8OulIodhcs6_APNtRsA.png" alt="Centered Image" width="650">

#### <br>**Schauen wir den programmierten Algorithmus als Pseudocode genauer an.**

Wenn der Computer einen Zug macht, passiert Folgendes:

1. Die Funktion `find_best_move` wird aufgerufen.<br><br>
2. Für jeden möglichen Zug auf dem aktuellen Spielfeld:
   - Der Zug wird hypothetisch ausgeführt.
   - Die `minimax` Funktion wird aufgerufen, um den Wert dieses Zuges zu ermitteln.
   - Der Zug wird rückgängig gemacht.<br><br>
3. In der `minimax` Funktion:
   - Wenn die maximale Tiefe erreicht ist oder das Spiel zu Ende ist, wird der aktuelle Spielstand bewertet.
   - Ansonsten werden rekursiv alle möglichen Folgezüge untersucht:
     - Für den maximierenden Spieler (Computer) wird der höchste Wert gewählt.
     - Für den minimierenden Spieler (Gegner) wird der niedrigste Wert gewählt.<br><br>
4. Der Prozess wiederholt sich rekursiv, wobei abwechselnd maximiert und minimiert wird, bis die Blätter des Spielbaums erreicht sind.<br><br>
5. Die Werte werden den Baum hinauf propagiert, wobei jeder Spieler (Max oder Min) die für ihn optimale Wahl trifft.<br><br>
6. Am Ende wählt `finde_besten_zug` den Zug mit dem höchsten Wert für den Computer aus.<br>


```python
def minimax(board, depth, is_maximizing):
    # Minimax-Algorithmus für die TicTacToe-KI
    scores = {'X': 1, 'O': -1, 'Tie': 0}
    
    # Basisfall: Überprüfe, ob das Spiel beendet ist
    winner = check_winner(board)
    if winner:
        return scores[winner]
    if is_full(board):
        return scores['Tie']
    
    if is_maximizing:
        # Maximierender Spieler (Computer, 'X')
        best_score = float('-inf')
        for i in range(3):
            for j in range(3):
                if board[i][j] == ' ':
                    # Probiere diesen Zug
                    board[i][j] = 'X'
                    # Rekursiver Aufruf für den minimierenden Spieler
                    score = minimax(board, depth + 1, False)
                    # Mache den Zug rückgängig
                    board[i][j] = ' '
                    # Aktualisiere den besten Score
                    best_score = max(score, best_score)
        return best_score
    else:
        # Minimierender Spieler (Mensch, 'O')
        best_score = float('inf')
        for i in range(3):
            for j in range(3):
                if board[i][j] == ' ':
                    # Probiere diesen Zug
                    board[i][j] = 'O'
                    # Rekursiver Aufruf für den maximierenden Spieler
                    score = minimax(board, depth + 1, True)
                    # Mache den Zug rückgängig
                    board[i][j] = ' '
                    # Aktualisiere den besten Score
                    best_score = min(score, best_score)
        return best_score

def find_best_move(board):
    # Finde den besten Zug für den Computer
    best_score = float('-inf')
    best_move = None
    for i in range(3):
        for j in range(3):
            if board[i][j] == ' ':
                # Probiere jeden möglichen Zug
                board[i][j] = 'X'
                # Bewerte diesen Zug mit Minimax
                score = minimax(board, 0, False)
                # Mache den Zug rückgängig
                board[i][j] = ' '
                # Aktualisiere den besten Zug, wenn ein besserer gefunden wurde
                if score > best_score:
                    best_score = score
                    best_move = (i, j)
    return best_move
```

```python
import random

def print_board(board):
    for row in board:
        print(" | ".join(row))
        print("---------")

def check_winner(board):
    # Überprüfe Reihen, Spalten und Diagonalen auf einen Gewinner
    for i in range(3):
        if board[i][0] == board[i][1] == board[i][2] != ' ':
            return board[i][0]
        if board[0][i] == board[1][i] == board[2][i] != ' ':
            return board[0][i]
    if board[0][0] == board[1][1] == board[2][2] != ' ':
        return board[0][0]
    if board[0][2] == board[1][1] == board[2][0] != ' ':
        return board[0][2]
    return None

def is_full(board):
    # Überprüfe, ob das Brett voll ist (Unentschieden)
    return all(cell != ' ' for row in board for cell in row)

values = []

def minimax(board, depth, is_maximizing):
    # Minimax-Algorithmus für die TicTacToe-KI
    scores = {'X': 1, 'O': -1, 'Tie': 0}
    
    # Basisfall: Überprüfe, ob das Spiel beendet ist
    winner = check_winner(board)
    if winner:
        return scores[winner]
    if is_full(board):
        return scores['Tie']
    
    if is_maximizing:
        # Maximierender Spieler (Computer, 'X')
        best_score = float('-inf')
        for i in range(3):
            for j in range(3):
                if board[i][j] == ' ':
                    # Probiere diesen Zug
                    board[i][j] = 'X'
                    # Rekursiver Aufruf für den minimierenden Spieler
                    score = minimax(board, depth + 1, False)
                    # Mache den Zug rückgängig
                    board[i][j] = ' '
                    # Aktualisiere den besten Score
                    best_score = max(score, best_score)
        values.append(best_score)
        return best_score
    else:
        # Minimierender Spieler (Mensch, 'O')
        best_score = float('inf')
        for i in range(3):
            for j in range(3):
                if board[i][j] == ' ':
                    # Probiere diesen Zug
                    board[i][j] = 'O'
                    # Rekursiver Aufruf für den maximierenden Spieler
                    score = minimax(board, depth + 1, True)
                    # Mache den Zug rückgängig
                    board[i][j] = ' '
                    # Aktualisiere den besten Score
                    best_score = min(score, best_score)
        values.append(best_score)
        return best_score

def find_best_move(board):
    # Finde den besten Zug für den Computer
    best_score = float('-inf')
    best_move = None
    for i in range(3):
        for j in range(3):
            if board[i][j] == ' ':
                # Probiere jeden möglichen Zug
                board[i][j] = 'X'
                # Bewerte diesen Zug mit Minimax
                score = minimax(board, 0, False)
                # Mache den Zug rückgängig
                board[i][j] = ' '
                # Aktualisiere den besten Zug, wenn ein besserer gefunden wurde
                if score > best_score:
                    best_score = score
                    best_move = (i, j)
    # Speichere alle gesammelten Werte in eine Textdatei
    with open('minimax_values.txt', 'w') as f:
        for val in values:
            f.write(str(val) + '\n')
    return best_move


def player_move(board):
    while True:
        try:
            row = int(input("Geben Sie die Zeile ein (0-2): "))
            col = int(input("Geben Sie die Spalte ein (0-2): "))
            if 0 <= row <= 2 and 0 <= col <= 2 and board[row][col] == ' ':
                return row, col
            else:
                print("Ungültiger Zug. Versuchen Sie es erneut.")
        except ValueError:
            print("Bitte geben Sie gültige Zahlen ein.")

def play_game():
    board = [[' ' for _ in range(3)] for _ in range(3)]
    current_player = random.choice(['Spieler', 'Computer'])
    
    while True:
        print_board(board)
        if current_player == 'Spieler':
            print("Ihr Zug (O):")
            row, col = player_move(board)
            board[row][col] = 'O'
        else:
            print("Computer ist am Zug (X)...")
            row, col = find_best_move(board)
            board[row][col] = 'X'
        
        winner = check_winner(board)
        if winner:
            print_board(board)
            print(f"{winner} hat gewonnen!")
            break
        elif is_full(board):
            print_board(board)
            print("Unentschieden!")
            break
        
        current_player = 'Spieler' if current_player == 'Computer' else 'Computer'

if __name__ == "__main__":
    play_game()
```
