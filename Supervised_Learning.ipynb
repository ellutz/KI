Im Buch wird eine einfache Vorhersage für das Tiergwicht anhand der Tatzengrösse getroffen. Das ist eine lineare Regression, die der Zielwert nur anhand von einem Merkmal berechnet wird! Die untenstehenden Daten entsprechen nicht dem Buch, greifen aber die Idee nochmals auf.

<p>
  Hinweis: Der eingesetzte Wert wird im Python-Code nicht "von Hand berechnet", sondern mit Hilfe einer Bibliothek (vorgefertige Codes/Funktionen). In diesem Fall übernimmt <code>numpy</code> diese Aufgabe. Mit <code>np.polyfit(x, y, 1)</code> wird automatisch die beste Gerade zu den Daten bestimmt. Dabei werden Steigung <em>m</em> und Achsenabschnitt <em>b</em> so gewählt, dass die Abweichungen zwischen den Datenpunkten und der Geraden möglichst klein sind. Anschliessend kann für einen neuen Wert die Vorhersage mit der Geradengleichung berechnet werden.
</p>

<p>
  Im Tatzengrössen-Beispiel setzen wir nun <code>x_new = 40</code> ein. Das bedeutet: Für eine Tatzengrösse von 40 cm soll das zugehörige Tiergewicht geschätzt werden. Der vorhergesagte Wert ergibt sich dann aus der Gleichung <code>y_new = m * x_new + b</code>.
</p>
<br>

```python
import numpy as np
import matplotlib.pyplot as plt

# Daten
x = np.array([18, 22, 26, 30, 34, 38, 42])  # Tatzengrösse

# mehr Streuung
y = np.array([13.8, 15.6, 21.0, 30.4, 32.8, 42.2, 44.0])  # Tiergewicht

# lineare Regression
m, b = np.polyfit(x, y, 1)

# Gerade
x_line = np.linspace(min(x), max(x), 200)
y_line = m * x_line + b

# Vorhersagepunkt (roter Punkt auf Gerade)
x_new = 40
y_new = m * x_new + b

# Plot
fig, ax = plt.subplots(figsize=(5.2, 4.2))

ax.scatter(x, y, color="#2a9d8f", s=28, zorder=3)
ax.plot(x_line, y_line, color="black", linewidth=1.5)

ax.scatter([x_new], [y_new], color="#d62828", s=32, zorder=4)
ax.axvline(x_new, color="#d62828", linestyle=":", linewidth=1)
ax.axhline(y_new, color="#d62828", linestyle=":", linewidth=1)

ax.set_xlabel("Tatzengrösse", fontsize=10)
ax.set_ylabel("Tiergewicht", fontsize=10)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.set_xlim(16, 44)
ax.set_ylim(10, 50)

ax.text(12.3, 12.0, "leicht", fontsize=8)
ax.text(12.3, 47.0, "schwer", fontsize=8)
ax.text(16.5, 8.4, "klein", fontsize=8)
ax.text(41.8, 8.4, "gross", fontsize=8)

plt.tight_layout()
plt.show()
```


### 1.3 Multiple Lineare Regression

Die multiple lineare Regression ist ein Verfahren, mit dem man eine Grösse vorhersagen kann, indem man mehrere Einflussgrössen gleichzeitig betrachtet. Dadurch werden Vorhersagen genauer, wie das Beispiel der Honigmenge zeigen soll. Bei der multiplen linearen Regression werden mehrere Faktoren zusammen einbezogen, etwa Blütenangebot, Frühlingstemperatur und Anzahl sonniger Tage. Der Computer sucht dann nach einer möglichst guten mathematischen Beziehung, damit die Vorhersage möglichst genau wird. Man kann sich das so vorstellen: Nicht nur ein Faktor bestimmt das Ergebnis, sondern mehrere wirken gemeinsam zusammen. Das ist im Alltag oft realistischer, weil viele Dinge von verschiedenen Einflüssen abhängen. Ein gutes Beispiel aus dem Bereich Machine Learning (Vorhersagen) ist die Vorhersage von Wohnungspreisen. Dabei spielen nicht nur die Grösse einer Wohnung eine Rolle, sondern auch die Anzahl Zimmer, die Lage oder das Stockwerk. Genau dafür ist die multiple lineare Regression sinnvoll: Sie hilft, mehrere Einflussfaktoren gleichzeitig zu berücksichtigen und dadurch bessere Vorhersagen zu machen.
```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Daten (Einflussgrössen für Honigmenge aus oberer Tabelle)
X = np.array([
    [40, 12.0, 16],
    [34, 10.9, 12],
    [48, 13.4, 19],
    [29, 9.7, 10],
    [45, 12.6, 17],
    [53, 13.8, 22],
    [37, 11.3, 14],
    [58, 14.5, 24],
    [32, 10.5, 11],
    [47, 13.0, 18],
    [41, 11.9, 15],
    [30, 9.9, 9],
    [55, 13.9, 21],
    [43, 12.3, 16],
    [50, 14.1, 23]
])

# Daten (Tatsächliche Honigmenge aus oberer Tabelle)
y = np.array([17.1, 13.8, 20.5, 11.6, 18.9, 23.1, 15.4, 25.0, 13.0, 19.6, 17.0, 10.7, 24.0, 18.2, 21.9])

# Modell erstellen und trainieren
modell = LinearRegression()
modell.fit(X, y)

# Koeffizienten und Achsenabschnitt
print("Achsenabschnitt:", modell.intercept_)
print("Koeffizienten:", modell.coef_)

# Vorhersage für ein neues Jahr
neue_daten = np.array([[43, 12.5, 18]])
vorhersage = modell.predict(neue_daten)

print("Vorhergesagte Honigmenge:", vorhersage[0])

# -------------------------------------------------
# Plot: Tatsächliche Werte vs. Vorhersagen des Modells
# -------------------------------------------------
y_pred = modell.predict(X)

plt.figure(figsize=(6, 5))
plt.scatter(y, y_pred, color="#2a9d8f", s=35)

# Diagonale = perfekte Vorhersage
min_wert = min(y.min(), y_pred.min())
max_wert = max(y.max(), y_pred.max())
plt.plot([min_wert, max_wert], [min_wert, max_wert], color="black", linewidth=1.5)

# Vorhersagepunkt für neue Daten markieren
plt.scatter(vorhersage[0], vorhersage[0], color="#d62828", s=45, zorder=3)

plt.xlabel("Tatsächliche Honigmenge (kg)")
plt.ylabel("Vorhergesagte Honigmenge (kg)")
plt.title("Multiple lineare Regression")
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

<br>Zu **modell.fit(X,y)** im Code: Mit der „besten Kombination aus Parametern“ sind die Werte für \(b_0, b_1, b_2\) und \(b_3\) gemeint, bei denen die Vorhersagen des Modells möglichst gut zu den echten Daten passen. Gut ist eine Kombination dann, wenn die Unterschiede zwischen vorhergesagten und tatsächlichen Werten möglichst klein sind. Der Computer sucht also die Parameter, mit denen die Honigmenge aus Blütenangebot, Frühlingstemperatur und sonnigen Tagen am genauesten beschrieben werden kann.

## 2. k-Nächste-Nachbarn - Klassifizierung

<br>Beim k-Nächste-Nachbarn-Verfahren, kurz k-NN, wird ein neuer Datenpunkt nicht mit einer fest gelernten Regel eingeordnet, sondern mit den bereits bekannten Beispielen verglichen. Das Modell schaut also, welche gespeicherten Datenpunkte dem neuen Punkt am ähnlichsten sind. Die Ähnlichkeit wird über den Abstand zwischen den Punkten bestimmt: Je kleiner der Unterschied, desto ähnlicher sind sie. Der Parameter \(k\) gibt an, wie viele Nachbarn betrachtet werden. Danach wird die Klasse gewählt, die unter diesen nächsten Nachbarn am häufigsten vorkommt. Genau dieses Prinzip haben wir in Kapitel 3 bereits mit den Bienenbildern von Hand nachgestellt. Jetzt setzen wir es in Python um und lassen den Computer die ähnlichen Beispiele suchen und die Zuordnung automatisch machen.

Links in der Dateiübersicht finden Sie eine datei *bienen_arte.csv*. Darin sind Beispielmerkmale, angelehnt an der Aufgabenstellung bei zu Bienenbildern. Schauen Sie die Mermkale kurz an. Diese Datei wird anschliessen im Python-Code verwenden.
<br>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.neighbors import KNeighborsClassifier

# CSV laden
daten = pd.read_csv("bienen_arten.csv")

# Merkmale und Zielvariable
y = daten["Art"]
X = daten[["Farbe", "Körpergrösse", "Flügelspanne", "Behaarung", "Körperform"]].copy()

# Kategorien in Zahlen umwandeln
label_farbe = LabelEncoder()
label_behaarung = LabelEncoder()
label_körperform = LabelEncoder()

X["Farbe"] = label_farbe.fit_transform(X["Farbe"])
X["Behaarung"] = label_behaarung.fit_transform(X["Behaarung"])
X["Körperform"] = label_körperform.fit_transform(X["Körperform"])

# Skalieren
scaler = StandardScaler()
X_skaliert = scaler.fit_transform(X)

# k-NN trainieren
k = 3
modell = KNeighborsClassifier(n_neighbors=k)
modell.fit(X_skaliert, y)

# Neue Biene
neue_biene = pd.DataFrame([{
    "Farbe": "gelb-braun",
    "Körpergrösse": 5.2,
    "Flügelspanne": 9.8,
    "Behaarung": "mittel",
    "Körperform": "rundlich"
}])

neue_biene["Farbe"] = label_farbe.transform(neue_biene["Farbe"])
neue_biene["Behaarung"] = label_behaarung.transform(neue_biene["Behaarung"])
neue_biene["Körperform"] = label_körperform.transform(neue_biene["Körperform"])
neue_biene_skaliert = scaler.transform(neue_biene)

# Vorhersage
vorhersage = modell.predict(neue_biene_skaliert)
print("Vorhergesagte Bienenart:", vorhersage[0])

# Nächste Nachbarn bestimmen
abstaende = np.linalg.norm(X_skaliert - neue_biene_skaliert, axis=1)
nachbarn_index = np.argsort(abstaende)[:k]

# Plot
farben = {
    "Westliche Honigbiene": "gold",
    "Sandbiene": "black",
    "Holzbiene": "purple",
    "Wespenbiene": "red",
    "Blattschneiderbiene": "green"
}

plt.figure(figsize=(8, 6))

for art in daten["Art"].unique():
    auswahl = daten[daten["Art"] == art]
    plt.scatter(
        auswahl["Körpergrösse"],
        auswahl["Flügelspanne"],
        label=art,
        color=farben.get(art, "gray"),
        s=45
    )

# Neue Biene markieren
plt.scatter(
    neue_biene["Körpergrösse"],
    neue_biene["Flügelspanne"],
    color="blue",
    s=140,
    marker="X",
    label="Neue Biene"
)

# Abstandsmarken zu den k nächsten Nachbarn
for i in nachbarn_index:
    plt.plot(
        [neue_biene["Körpergrösse"].iloc[0], daten["Körpergrösse"].iloc[i]],
        [neue_biene["Flügelspanne"].iloc[0], daten["Flügelspanne"].iloc[i]],
        color="gray",
        linestyle="--",
        linewidth=1
    )

plt.xlabel("Körpergrösse")
plt.ylabel("Flügelspanne")
plt.title(f"k-Nächste-Nachbarn mit k = {k}")
plt.legend()
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```
