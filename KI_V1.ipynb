#### Minmax
Der Minmax-Algorithmus ist ein Entscheidungsalgorithmus, der häufig in Spielen wie Schach oder Tic-Tac-Toe verwendet wird, um optimale Züge zu berechnen. Er wird eingesetzt, wenn zwei Spieler gegeneinander spielen und beide versuchen, das beste Ergebnis für sich zu erzielen. Die Grundidee des Algorithmus ist, dass jeder Spieler versucht, seinen eigenen Gewinn zu maximieren, während er gleichzeitig den Gegner daran hindert, zu gewinnen.

#### <br>**Schauen wir den programmierten Algorithmus als Pseudocode genauer an.**

Wenn der Computer einen Zug macht, passiert Folgendes:

1. Die Funktion `find_best_move` wird aufgerufen.<br><br>
2. Für jeden möglichen Zug auf dem aktuellen Spielfeld:
   - Der Zug wird hypothetisch ausgeführt.
   - Die `minmax` Funktion wird aufgerufen, um den Wert dieses Zuges zu ermitteln.
   - Der Zug wird rückgängig gemacht.<br><br>
3. In der `minmax` Funktion:
   - Wenn die maximale Tiefe erreicht ist oder das Spiel zu Ende ist, wird der aktuelle Spielstand bewertet.
   - Ansonsten werden rekursiv alle möglichen Folgezüge untersucht:
     - Für den maximierenden Spieler (Computer) wird der höchste Wert gewählt.
     - Für den minimierenden Spieler (Gegner) wird der niedrigste Wert gewählt.<br><br>
4. Der Prozess wiederholt sich rekursiv, wobei abwechselnd maximiert und minimiert wird, bis die Blätter des Spielbaums erreicht sind.<br><br>
5. Die Werte werden den Baum hinauf propagiert, wobei jeder Spieler (Max oder Min) die für ihn optimale Wahl trifft.<br><br>
6. Am Ende wählt `finde_best_zug` den Zug mit dem höchsten Wert für den Computer aus.<br>


```python
def minmax(board, depth, is_maximizing):
    # Minmax-Algorithmus für die TicTacToe-KI
    scores = {'X': 1, 'O': -1, 'Tie': 0}
    
    # Basisfall: Überprüfe, ob das Spiel beendet ist
    winner = check_winner(board)
    if winner:
        return scores[winner]
    if is_full(board):
        return scores['Tie']
    
    if is_maximizing:
        # Maximierender Spieler (Computer, 'X')
        best_score = float('-inf')
        for i in range(3):
            for j in range(3):
                if board[i][j] == ' ':
                    # Probiere diesen Zug
                    board[i][j] = 'X'
                    # Rekursiver Aufruf für den minimierenden Spieler
                    score = minmax(board, depth + 1, False)
                    # Mache den Zug rückgängig
                    board[i][j] = ' '
                    # Aktualisiere den besten Score
                    best_score = max(score, best_score)
        return best_score
    else:
        # Minimierender Spieler (Mensch, 'O')
        best_score = float('inf')
        for i in range(3):
            for j in range(3):
                if board[i][j] == ' ':
                    # Probiere diesen Zug
                    board[i][j] = 'O'
                    # Rekursiver Aufruf für den maximierenden Spieler
                    score = minmax(board, depth + 1, True)
                    # Mache den Zug rückgängig
                    board[i][j] = ' '
                    # Aktualisiere den besten Score
                    best_score = min(score, best_score)
        return best_score

def find_best_move(board):
    # Finde den besten Zug für den Computer
    best_score = float('-inf')
    best_move = None
    for i in range(3):
        for j in range(3):
            if board[i][j] == ' ':
                # Probiere jeden möglichen Zug
                board[i][j] = 'X'
                # Bewerte diesen Zug mit Minmax
                score = minmax(board, 0, False)
                # Mache den Zug rückgängig
                board[i][j] = ' '
                # Aktualisiere den besten Zug, wenn ein besserer gefunden wurde
                if score > best_score:
                    best_score = score
                    best_move = (i, j)
    return best_move
```

```python
import random

def print_board(board):
    for row in board:
        print(" | ".join(row))
        print("---------")

def check_winner(board):
    # Überprüfe Reihen, Spalten und Diagonalen auf einen Gewinner
    for i in range(3):
        if board[i][0] == board[i][1] == board[i][2] != ' ':
            return board[i][0]
        if board[0][i] == board[1][i] == board[2][i] != ' ':
            return board[0][i]
    if board[0][0] == board[1][1] == board[2][2] != ' ':
        return board[0][0]
    if board[0][2] == board[1][1] == board[2][0] != ' ':
        return board[0][2]
    return None

def is_full(board):
    # Überprüfe, ob das Brett voll ist (Unentschieden)
    return all(cell != ' ' for row in board for cell in row)

values = []

def minmax(board, depth, is_maximizing):
    # Minmax-Algorithmus für die TicTacToe-KI
    scores = {'X': 1, 'O': -1, 'Tie': 0}
    
    # Basisfall: Überprüfe, ob das Spiel beendet ist
    winner = check_winner(board)
    if winner:
        return scores[winner]
    if is_full(board):
        return scores['Tie']
    
    if is_maximizing:
        # Maximierender Spieler (Computer, 'X')
        best_score = float('-inf')
        for i in range(3):
            for j in range(3):
                if board[i][j] == ' ':
                    # Probiere diesen Zug
                    board[i][j] = 'X'
                    # Rekursiver Aufruf für den minimierenden Spieler
                    score = minmax(board, depth + 1, False)
                    # Mache den Zug rückgängig
                    board[i][j] = ' '
                    # Aktualisiere den besten Score
                    best_score = max(score, best_score)
        values.append(best_score)
        return best_score
    else:
        # Minimierender Spieler (Mensch, 'O')
        best_score = float('inf')
        for i in range(3):
            for j in range(3):
                if board[i][j] == ' ':
                    # Probiere diesen Zug
                    board[i][j] = 'O'
                    # Rekursiver Aufruf für den maximierenden Spieler
                    score = minmax(board, depth + 1, True)
                    # Mache den Zug rückgängig
                    board[i][j] = ' '
                    # Aktualisiere den besten Score
                    best_score = min(score, best_score)
        values.append(best_score)
        return best_score

def find_best_move(board):
    # Finde den besten Zug für den Computer
    best_score = float('-inf')
    best_move = None
    for i in range(3):
        for j in range(3):
            if board[i][j] == ' ':
                # Probiere jeden möglichen Zug
                board[i][j] = 'X'
                # Bewerte diesen Zug mit Minmax
                score = minmax(board, 0, False)
                # Mache den Zug rückgängig
                board[i][j] = ' '
                # Aktualisiere den besten Zug, wenn ein besserer gefunden wurde
                if score > best_score:
                    best_score = score
                    best_move = (i, j)
    # Speichere alle gesammelten Werte in eine Textdatei
    with open('minmax_values.txt', 'w') as f:
        for val in values:
            f.write(str(val) + '\n')
    return best_move


def player_move(board):
    while True:
        try:
            row = int(input("Geben Sie die Zeile ein (0-2): "))
            col = int(input("Geben Sie die Spalte ein (0-2): "))
            if 0 <= row <= 2 and 0 <= col <= 2 and board[row][col] == ' ':
                return row, col
            else:
                print("Ungültiger Zug. Versuchen Sie es erneut.")
        except ValueError:
            print("Bitte geben Sie gültige Zahlen ein.")

def play_game():
    board = [[' ' for _ in range(3)] for _ in range(3)]
    current_player = random.choice(['Spieler', 'Computer'])
    
    while True:
        print_board(board)
        if current_player == 'Spieler':
            print("Ihr Zug (O):")
            row, col = player_move(board)
            board[row][col] = 'O'
        else:
            print("Computer ist am Zug (X)...")
            row, col = find_best_move(board)
            board[row][col] = 'X'
        
        winner = check_winner(board)
        if winner:
            print_board(board)
            print(f"{winner} hat gewonnen!")
            break
        elif is_full(board):
            print_board(board)
            print("Unentschieden!")
            break
        
        current_player = 'Spieler' if current_player == 'Computer' else 'Computer'

if __name__ == "__main__":
    play_game()
```

#### AlphaGo (zero)
Die KI lernt Tic-Tac-Toe, ohne dass ihr ein Mensch eine Strategie oder Bewertung vorgibt. Sie kennt nur die Spielregeln und spielt tausende Partien gegen sich selbst. Nach jeder Partie merkt sie sich in einer Tabelle, welche Stellungen zum Sieg, zum Unentschieden oder zur Niederlage geführt haben. Im Spiel gegen den Menschen wählt sie jeweils den Zug mit der bisher höchsten Erfolgsquote und wird so allein durch Erfahrung immer stärker, wie AlphaGo Zero, jedoch mit einer Tabelle statt eines neuronalen Netzes und ohne Suchbaum (MCTS).

```python
import random

def print_board(board):
    for row in board:
        print(" | ".join(row))
        print("---------")

def check_winner(board):
    # Überprüfe Reihen, Spalten und Diagonalen auf einen Gewinner
    for i in range(3):
        if board[i][0] == board[i][1] == board[i][2] != ' ':
            return board[i][0]
        if board[0][i] == board[1][i] == board[2][i] != ' ':
            return board[0][i]
    if board[0][0] == board[1][1] == board[2][2] != ' ':
        return board[0][0]
    if board[0][2] == board[1][1] == board[2][0] != ' ':
        return board[0][2]
    return None

def is_full(board):
    # Überprüfe, ob das Brett voll ist (Unentschieden)
    return all(cell != ' ' for row in board for cell in row)

# Gedaechtnis statt Minmax
gedaechtnis = {}   # (Stellung, Spieler) -> [Punkte, Anzahl Partien]

def stellung(board):
    return "".join(cell for row in board for cell in row)

def wert(board, spieler):
    punkte, anzahl = gedaechtnis.get((stellung(board), spieler), [0, 0])
    return 0.5 if anzahl == 0 else punkte / anzahl   # unbekannt = 50 %

def freie_felder(board):
    return [(i, j) for i in range(3) for j in range(3) if board[i][j] == ' ']

def find_best_move(board, spieler='X', neugier=0.0):
    # Finde den besten Zug: Erfahrung nachschlagen statt vorausrechnen
    felder = freie_felder(board)
    if random.random() < neugier:              # manchmal etwas Neues ausprobieren
        return random.choice(felder)
    bewertungen = {}
    for (i, j) in felder:
        board[i][j] = spieler                  # Probiere diesen Zug
        bewertungen[(i, j)] = wert(board, spieler)   # Erfahrung nachschlagen
        board[i][j] = ' '                      # Mache den Zug rueckgaengig
    bester = max(bewertungen.values())
    return random.choice([f for f in bewertungen if bewertungen[f] == bester])

# Selbstspiel und Lernen
def selbstspiel():
    board = [[' ' for _ in range(3)] for _ in range(3)]
    spieler = random.choice(['X', 'O'])
    verlauf = []
    while True:
        i, j = find_best_move(board, spieler, neugier=0.2)
        board[i][j] = spieler
        verlauf.append((stellung(board), spieler))
        winner = check_winner(board)
        if winner or is_full(board):
            return winner, verlauf
        spieler = 'O' if spieler == 'X' else 'X'

def lerne(winner, verlauf):
    for st, spieler in verlauf:
        if winner == spieler:  punkte = 1      # gewonnen
        elif winner is None:   punkte = 0.5    # unentschieden
        else:                  punkte = 0      # verloren
        eintrag = gedaechtnis.setdefault((st, spieler), [0, 0])
        eintrag[0] += punkte
        eintrag[1] += 1

def trainiere(anzahl):
    for _ in range(anzahl):
        lerne(*selbstspiel())

def speichere_gedaechtnis(dateiname='gedaechtnis.txt'):
    # Sortiert: haeufigste Stellungen zuerst
    eintraege = sorted(gedaechtnis.items(), key=lambda e: e[1][1], reverse=True)
    with open(dateiname, 'w', encoding='utf-8') as f:
        f.write(f"Gedaechtnis der KI: {len(gedaechtnis)} Stellungen\n")
        f.write("Brett nach dem Zug, Zeilen mit | getrennt, . = leer\n")
        f.write("Erfolgsquote: Sieg = 1, Remis = 0.5, Niederlage = 0\n\n")
        for (st, spieler), (punkte, anzahl) in eintraege:
            b = st.replace(' ', '.')
            brett = b[0:3] + "|" + b[3:6] + "|" + b[6:9]
            f.write(f"{brett}   Zug von {spieler}   Erfolgsquote {punkte/anzahl:5.0%}   ({anzahl} Partien)\n")

# Unverändert
def player_move(board):
    while True:
        try:
            row = int(input("Geben Sie die Zeile ein (0-2): "))
            col = int(input("Geben Sie die Spalte ein (0-2): "))
            if 0 <= row <= 2 and 0 <= col <= 2 and board[row][col] == ' ':
                return row, col
            else:
                print("Ungültiger Zug. Versuchen Sie es erneut.")
        except ValueError:
            print("Bitte geben Sie gültige Zahlen ein.")

def play_game():
    board = [[' ' for _ in range(3)] for _ in range(3)]
    current_player = random.choice(['Spieler', 'Computer'])
    
    while True:
        print_board(board)
        if current_player == 'Spieler':
            print("Ihr Zug (O):")
            row, col = player_move(board)
            board[row][col] = 'O'
        else:
            print("Computer ist am Zug (X)...")
            row, col = find_best_move(board)
            board[row][col] = 'X'
        
        winner = check_winner(board)
        if winner:
            print_board(board)
            print(f"{winner} hat gewonnen!")
            break
        elif is_full(board):
            print_board(board)
            print("Unentschieden!")
            break
        
        current_player = 'Spieler' if current_player == 'Computer' else 'Computer'

if __name__ == "__main__":
    anzahl = int(input("Wie viele Trainingspartien? (z.B. 0, 1000, 50000): "))
    trainiere(anzahl)
    speichere_gedaechtnis()
    print(f"Die KI kennt jetzt {len(gedaechtnis)} Stellungen.")
    print("Gedaechtnis gespeichert in gedaechtnis.txt\n")
    play_game()
```
